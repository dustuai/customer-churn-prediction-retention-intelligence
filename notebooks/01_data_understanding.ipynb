# 01 — Data Understanding

Audit dimensions, types, missing values, duplicates and target balance.

```python
from pathlib import Path
import sys, pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path.cwd().resolve()
if not (ROOT/"src").exists(): ROOT = ROOT.parent
sys.path.append(str(ROOT))
DATA = ROOT/"data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"
assert DATA.exists(), f"Download the dataset and place it at {DATA}"

df=pd.read_csv(DATA)
print(df.shape)
display(df.head())
display(pd.DataFrame({'dtype':df.dtypes.astype(str),'missing':df.isna().sum(),'unique':df.nunique()}).sort_values('missing',ascending=False))
print('Duplicate rows:',df.duplicated().sum())
print('Duplicate IDs:',df.customerID.duplicated().sum())
display(df.Churn.value_counts().to_frame('count'))
```
